```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

#Translation Model
translation_name = "facebook/nllb-200-distilled-600M"
translation_tokenizer = AutoTokenizer.from_pretrained(translation_name)
translation_model = AutoModelForSeq2SeqLM.from_pretrained(translation_name)

#code in facebook model for hebrew
hebrew_code = "heb_Hebr"

#Test
text_test1 = "It`s a sunny day outside isn`t it?"
#Tokenize the sentence , pt stans for PyTorch framework
inputs = translation_tokenizer(text_test1,return_tensors = "pt")
#BOS - Beginning of Sentence - this line is like the dictionary
forced_bos_token_id = translation_tokenizer.convert_tokens_to_ids(hebrew_code)

#this translates the tokens into hebrew tokens via the forced_bos_token (Hebrew Dictionary)
translated_tokens = translation_model.generate(
    **inputs,
    forced_bos_token_id=forced_bos_token_id,
    max_length=1000
)

# Translate the tokens into strings so we can read it , translated_tokens[0] the first token where it starts
# skip_special_tokens to skip all the <s> that the model adds so we have a readable text
translation = translation_tokenizer.decode(translated_tokens[0], skip_special_tokens=True)
print("Translation:", translation)

```

```python
from ollama import Client

#Local ollama client running
client = Client(host='http://localhost:11434')


#I call it ph3mini and not phi3:mini-4k-instruct because i have created a new custom model via gguf file of phi3:mini... in ollama
#and now if i want to use that model i have to use the custom name i called phi3mini
summary_model = "phi3mini"

#Trying out system promprt and temperature to future use case in the little project
system_prompt = "You are a proffesional event planner, you give ideas for partys and events, be proffesional and give" \
"useful ideas and easy to implement , keep your answers short and to the point , show your answer organized by 5 bullet points"
prompt = "Give me an idea for a surprise party for my fiance in Israel"

#temperature controls the creativity - low less creative - high more creative
temperature = 0.8


response = client.chat(
    model = summary_model,
    options = {
        "temperature" : temperature
    },
    messages = [{'role' : "system", 'content' : system_prompt},
                {'role' : "user", 'content' : prompt}],
                stream = True)

print("Model Response in real time:")


for chunk in response:
    if 'content' in chunk['message']:
        print(chunk['message']['content'], end='', flush=True)


#print(f"Model response: {response['message']['content']}")

""" Got some intersting ideas for the birthday party :)"""
```
